Transforming Data (Feature Engineering)

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("raw_data.csv")
```

```python
df
```

```python
df2 = df.copy()
df2["tax"] =  df2["income"].apply(lambda x: "20%" if x>=50000 else "10%") 
df2
```

```python
#   map 
gender_map = {"Male": "M", "Female":"F", "Unknown" : "N"}
df2["gender"] = df2["gender"].map(gender_map)
df2
```

```python
df2 = df2.assign(new_income = df2["income"]*1.1)
```

```python
df2
```

```python
df2["country"] =   df2["country"].replace("USA","United states")
```

```python
df2
```

```python
df2.columns = ["Id","Name","Age","Country","Gender","Income", "Tax","New_income" ]
df2
```

```python
df2 = df2.rename(columns={"Income" : "Salary"})
df2
```

```python
df2.rename(index = {1: "First"})
```

```python
df2.sort_values("Salary")
sorted_df = df2.sort_values(["Salary", "Age"])
sorted_df
sorted_df.sort_index()
sorted_df.reset_index()
```
