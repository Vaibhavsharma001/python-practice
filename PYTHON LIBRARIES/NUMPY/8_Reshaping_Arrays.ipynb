Reshaping Arrays

```python
import numpy as np
```

```python
arr = np.random.randint(0, 50, size=(6,4))
```

```python
arr
```

```python
arr = arr.reshape(4, 6) 
```

```python
arr
```

```python
# collapse to 1D — always returns a copy
arr.flatten()      
```

```python
arr
```

```python
# swap any two axes, useful beyond 2D
arr.swapaxes(0, 1) 
```
