MISSING VALUE -EMPTY DATA HANDLING

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("raw_data.csv")
```

```python
df
```

```python
df.isnull()
```

```python
#isnull and isna gives same output
df.isna()
```

```python
#countno. of nan value
df.isnull().sum()
```

```python
#drops the row where it sees NAN value
df.dropna()
```

```python
#drops the column where it sees NAN value
df.dropna(axis=1)
```

```python
#replaces NAN value with 0
df.fillna(0)
```

```python
#select th column to fill the null values
df["age"].fillna(0)
```

```python
#age colunm ke average ko nul value me fill kro 
age_mean = df["age"].mean()
df["age"].fillna(age_mean)
```

```python
cleaned_data = df.copy()
age_mean = cleaned_data["age"].mean()
cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)
cleaned_data
```

```python
#ffill()---->forward fill
df.ffill()
```

```python
#bfill() ---->backward fill
df.bfill()
```
