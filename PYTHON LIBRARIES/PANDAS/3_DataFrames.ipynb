DataFrames

```python
import pandas as pd
import numpy as np
```

```python
# DataFrame in pandas - using dictionary
info = {
    "Name" : ["Vaibhav", "Aditi", "Romit"],
    "Marks" : [78, 99, 85],
    "Grade" : ['B', 'O', 'A']
}

df = pd.DataFrame(info)
print(df)
print(type(df))

print(df.index)        # row labels
print(df.columns)      # column labels
```

```python
# Dataframe using Lists
df = pd.DataFrame([["vaibhav",9],["vinay",8.5],["gaurav",6.9]], columns=["name", "cgpa"])
df
```

```python
# DataFrameusing Numpy array
np_arr = np.array([[1,2,3],[4,5,6]])
dg = pd.DataFrame(np_arr,columns=["a","b","c"])
dg
```
